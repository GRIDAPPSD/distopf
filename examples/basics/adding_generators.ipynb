```python
import distopf as opf
```

Run a power flow on the IEEE 123 bus system with no control variables.

```python
case = opf.create_case(opf.CASES_DIR / "csv" / "ieee123")
case.modify(v_max=1.1, v_min=0.95)
result = case.run_pf()
result.plot_network()
```

```python
case.add_generator("13", phases="ab", p=0.02, q=0.0)
case.control_variable="P"
result = case.run_pf()
result.plot_network()
```

```python
result.plot_voltages()
```

```python
result.plot_gens()
```

```python
case.add_capacitor(name="66", phases="ac", q=0.25)
result = case.run_pf()

result.plot_voltages()
```
